```python
from langchain_community.document_loaders import DirectoryLoader

# Initialize the DirectoryLoader with the path to the directory for text files
loader = DirectoryLoader("/content/data", glob="*.txt")

# Load the text data from the directory
dataset = loader.load()

for data in dataset:
  print("------------")
  print (data.page_content)
  print(data.metadata)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os
from google.colab import userdata


if "GEMINI_API_KEY" not in os.environ:
    os.environ["GEMINI_API_KEY"] = userdata.get('GEMINI_API_KEY')

# Initialize model for token counting
llm_for_token_count = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.7
)



if 'dataset' in locals() and len(dataset) > 0:
    # Calculate token count for each document in the dataset
    print("Token counts for each document using Gemini tokenizer:")
    for i, doc in enumerate(dataset):
        token_count = llm_for_token_count.get_num_tokens(doc.page_content)
        print(f"  Document {i+1} (Source: {doc.metadata.get('source', 'N/A')}): {token_count} tokens")
else:
    print("Dataset not found or is empty. Cannot count tokens.")
```

# Chunking the Text using **RecursiveCharacterTextSplitter**

```python
!pip install -U langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=0,
    length_function=len
)
```

```python
# Split the text of the second document in the dataset into chunks
chunks = text_splitter.split_text(dataset[1].page_content)

len(chunks)
```

```python
print(chunks[0])
print(len(chunks[0]))
print(chunks[1])
print(len(chunks[1]))
```

# Chunking the Text Using tiktoken Length Function

```python
# Reinitialize the text splitter witha chunk size of 150 and an overl;ap of 20, using tiktoken length function
import tiktoken
from langchain_text_splitters import RecursiveCharacterTextSplitter

enc = tiktoken.get_encoding("cl100k_base")

def tiktoken_len(text):
    return len(enc.encode(text))

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=20,
    length_function=tiktoken_len,
    separators=["\n\n", "\n", " ", ""]
)
```

```python
# Split the text of the second document into chunks
chunks = text_splitter.split_text(dataset[1].page_content)

# Number of chunks
len(chunks)
```

```python
print(chunks[0])
print(tiktoken_len(chunks[0]))
print(chunks[1])
print(tiktoken_len(chunks[1]))
```
