```python
import os
from google.colab import userdata
from pydantic import SecretStr # Import SecretStr for type checking

import warnings
warnings.filterwarnings('ignore')

# Safely extract the API key, handling both SecretStr and plain string types
api_key_value = userdata.get('GEMINI_API_KEY')

if isinstance(api_key_value, SecretStr):
    os.environ["GOOGLE_API_KEY"] = api_key_value.get_secret_value()
else:
    os.environ["GOOGLE_API_KEY"] = api_key_value
```

# Initialize Google Gemini LLM

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from google.colab import userdata
import os

# Initialize the GenAI model
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.7,
    # The API key will be automatically picked up from the GOOGLE_API_KEY environment variable.
)
```

# Initialize Embedding Model

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from google.colab import userdata
import os

# Initialize the embedding model
embedding_model = GoogleGenerativeAIEmbeddings(
    model="text-embedding-004", # Correctly updated to a widely supported Gemini embedding model
    # The API key will be automatically picked up from the GOOGLE_API_KEY environment variable.
)
```

# Load PDF Document

```python
from langchain_community.document_loaders import PyPDFLoader

# Load the PDF document
loader = PyPDFLoader("/content/codeprolk.pdf")

docs = loader.load()
```

```python
len(docs)
```

# Split Documents into Chunks

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Initialize the text splitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=20)

# Split the documents into chunks
documents = text_splitter.split_documents(docs)
```

```python
len(documents)
```

# Create Vector Store and Retriever

```python
from langchain_chroma import Chroma

# Create a vector store from the document chunks
vectorstore = Chroma.from_documents(documents=documents, embedding=embedding_model)
```

```python
# Create a retriever from the vector store
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})
```

# Define Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough

# Define a message template for the chatbot
message = """
Answer this question using the provided context only.

{question}

Context:
{context}

"""

# Create a chat prompt template from the message
prompt = ChatPromptTemplate.from_messages([("human", message)])
```

# Create Retrieval-Augmented Generation (RAG) chain

```python
from langchain_core.output_parsers import StrOutputParser

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

# Invoke RAG Chain with Example Questions

```python
response = chain.invoke("who is codeprolk")
print(response)
```

# Evaluate RAG Application

```python
import pandas as pd
test_data = pd.read_csv("/content/test_data.csv")
test_data
```

```python
questions = test_data ["question"].to_list()
ground_truth = test_data ["answer"].to_list()
```

```python
questions
```

```python
ground_truth
```

```python
data = {"question" : [], "answer": [], "contexts" : [], "ground_truth" : ground_truth}
```

```python
for query in questions:
  data ["question"].append(query)
  data["answer"].append(chain.invoke(query))
  data["contexts"].appnd([doc.page_content for doc in retriever.get_relevant_documents(query)])
```

```python
data
```

```python
# to evaluate from ragas
from datasets import Dataset

dataset = Dataset.from_dict(data)
```

```python
dataset
```

```python
dataset [0]
```

```python
from ragas.metrics import context_precision, context_recall, answer_relevancy, faithfuness
from ragas import evaluate

result = evaluate (
    dataset = dataset,
    metrics = [
        context_precision,
        context_recall,
        answer_relevancy,
        faithfuness],
)
```

```python
result
```

```python
results = results.to_pandas()
results
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

heatmap_data = results [['context_precision','context_recall','answer_relavancy','faithfulness']]

plt.figure(figsize=(10,8))
sns.heatmap(heatmap_data, annot=True,fmt=".2f", linwidths =.5)

plt.yticks(ticks = range (len(results['question'])), lables = results ['question'], rotation =0)
```
