```python
from langchain_community.document_loaders import DirectoryLoader

# Initialize the DirectoryLoader with tha path to the directory and a glob pattern for text files
loader = DirectoryLoader ("/content/drive/MyDrive/Colab Notebooks/data/txt_folder",glob = '**/*.txt')

# Load the text data from the directory
dataset = loader.load()

for data in dataset:
  print(".....................")
  print(data.page_content)
  print(data.metadata)
```

## Summary of Document Loading in LangChain

This notebook demonstrates how to load various types of documents into LangChain's `Document` format, which is essential for processing them with LangChain applications.

### Key Concepts Covered:

1.  **Text Files (`.txt`)**:
    *   **`TextLoader`**: Used to load individual `.txt` files. It extracts the raw text content from the file.
    *   **`DirectoryLoader`**: Used to load multiple text files from a specified directory. You can use `glob` patterns (e.g., `**/*.txt`) to filter for specific file types.

2.  **PDF Files (`.pdf`)**:
    *   **`PyPDFLoader`**: Used to load individual `.pdf` files. It extracts text from each page of the PDF.
    *   **`PyPDFDirectoryLoader`**: Used to load multiple PDF files from a directory, similar to `DirectoryLoader` for text files.

3.  **CSV Files (`.csv`)**:
    *   **`CSVLoader`**: Loads data from `.csv` files. Each row in the CSV is typically treated as a separate `Document`, with the content often being a concatenation of the row's values and metadata containing row-specific information.

4.  **HTML Files (`.html`)**:
    *   **`BSHTMLLoader`**: Utilizes BeautifulSoup to parse HTML files and extract clean text content, often including the title in the document's metadata.

### General Process:

*   **Import the Loader**: Start by importing the specific loader from `langchain_community.document_loaders`.
*   **Initialize the Loader**: Provide the file path or directory path to the loader.
*   **Load Data**: Call the `.load()` method on the initialized loader to retrieve a list of `Document` objects.
*   **Access Content/Metadata**: Each `Document` object has a `page_content` attribute (the extracted text) and a `metadata` attribute (a dictionary containing information like source path, page number, title, etc.).
